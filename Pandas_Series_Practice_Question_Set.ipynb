```python
# Question 29
import pandas as pd
from pandas import Series

# Create Series
sales = Series(
    [450, 750, 300, 900],
    index=["Pen", "Notebook", "Bag", "Laptop"]
)

print("Values:\n",sales.values)
print("\nIndex:\n", sales.index)
print("\nLaptop sales: \n", sales["Laptop"])
print("\nSales above 500:\n",sales[sales > 500])
print("Bag" in sales.index)

sales_dict = sales.to_dict()

print("\nDictionary:\n", sales_dict)
```
