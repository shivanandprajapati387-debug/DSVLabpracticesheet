```python
# Question 35
import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))

print("Marks:", marks)

student_avg = np.mean(marks, axis=1)
test_mean = np.mean(marks, axis=0)
test_std = np.std(marks, axis=0)

print("Average marks of each student:", student_avg)

print("\n Mean of each test:",test_mean)

print("\n Standard deviation of each test:",test_std)

passed_all = np.all(marks >= 40, axis=1)

print("\n Students passing all four tests:", np.sum(passed_all))

high_average = np.where(student_avg >= 75)[0] + 1

print("\n Students with average >= 75:", high_average)

most_variable_test = np.argmax(test_std) + 1

print("\n Test with greatest variation:", most_variable_test)
```

```python
# Question 36
import numpy as np

a = np.arange(21, 46).reshape(5, 5)

print("Original 5 x 5 matrix:\n",a)

central = a[1:4, 1:4]

print("\n Central 3 x 3 matrix: \n",central)

transpose = central.T

print("\n Transpose:\n",transpose)

product = np.matmul(central, transpose)

print("\n Matrix product:\n",product)

row_sums = np.sum(product, axis=1)

print("\n Row-wise sums:\n",row_sums)

maximum = np.max(product)

positions = np.argwhere(product == maximum)

print(" Maximum value:", maximum)
print("\n Positions of maximum (zero-based):", positions)
```

```python
# Question 37
import numpy as np

quantities = np.array([
    [10, 15, 20, 8],
    [12, 18, 15, 10],
    [20, 12, 25, 15],
    [18, 22, 10, 12],
    [25, 15, 18, 20],
    [16, 20, 22, 14]
])

original_prices = np.array([150, 240, 90, 320])

daily_revenue = np.sum(quantities * original_prices, axis=1)

product_revenue = np.sum(quantities * original_prices, axis=0)

print("Daily revenue:", daily_revenue)

print("\n Total revenue per product:",product_revenue)

highest_day = np.argmax(daily_revenue) + 1

print("\n Highest-revenue day:", highest_day)
print("\n Highest daily revenue:", np.max(daily_revenue))

total_quantity = np.sum(quantities, axis=0)

best_product = np.argmax(total_quantity) + 1

print("\n Total quantity per product:", total_quantity)

print("\n Best-selling product: Product", best_product)

new_prices = np.where(
    total_quantity > 100,
    original_prices * 0.90,
    original_prices
)

print("\n Original prices:",original_prices)

print("\n Reduced prices:",new_prices)

new_total_revenue = np.sum(quantities * new_prices)

print("\n Total revenue after price reduction:", new_total_revenue)
```

```python
# Question 38
import numpy as np

np.random.seed(42)

data = np.random.normal(50, 12, 100)

outside = np.sum((data < 30) | (data > 70))

cleaned = np.where(
    data < 30,
    30,
    np.where(data > 70, 70, data)
)

print("Original data:", data)

print("\n Number of observations outside 30-70:", outside)

print("\n Cleaned data:", cleaned)

print("\nOriginal statistics:")
print("\n Mean:", np.mean(data))
print("\n Variance:", np.var(data))
print("\n Standard deviation:", np.std(data))
print("\n Minimum:", np.min(data))
print("\n Maximum:", np.max(data))

print("\nCleaned statistics:")
print("\n Mean:", np.mean(cleaned))
print("\n Variance:", np.var(cleaned))
print("\n Standard deviation:", np.std(cleaned))
print("\n Minimum:", np.min(cleaned))
print("\n Maximum:", np.max(cleaned))
```

```python
# Question 39
import pandas as pd

branch_a = pd.Series({
    "Q1": 120,
    "Q2": 145,
    "Q4": 180
})

branch_b = pd.Series({
    "Q2": 110,
    "Q3": 135,
    "Q4": 155
})

branch_c = pd.Series({
    "Q1": 95,
    "Q3": 125,
    "Q4": 140
})

result = branch_a + branch_b + branch_c

print("Combined earnings using ordinary addition:")
print(result)

print("\nQuarters with missing combined earnings:")
print(result[result.isnull()])

common_quarters = (
    branch_a.index
    .intersection(branch_b.index)
    .intersection(branch_c.index)
)

print("\nQuarters common to all branches:")
print(common_quarters)

common_earnings = (
    branch_a[common_quarters]
    + branch_b[common_quarters]
    + branch_c[common_quarters]
)

print("\nCombined earnings for common quarters:")
print(common_earnings)

result_zero = (
    branch_a.add(branch_b, fill_value=0)
            .add(branch_c, fill_value=0)
)

print("\nCombined earnings treating missing records as zero:")
print(result_zero)
```

```python
# Question 40
import pandas as pd

scores = {
    "R101": 78,
    "R102": 65,
    "R104": 92,
    "R105": 48
}

required = [
    "R105",
    "R103",
    "R101",
    "R106",
    "R104",
    "R102"
]

s = pd.Series(scores, index=required)

print("Original Series:")
print(s)

print("\nMissing student records:")
print(s[s.isnull()])

print("\nValid scores of at least 70:")
valid_scores = s[s.notnull() & (s >= 70)]

print(valid_scores)

print("\nConvert filtered Series to dictionary:")
score_dict = valid_scores.to_dict()

print(score_dict)

print("\nReconstruct Series:")
new_series = pd.Series(score_dict)

print(new_series)

print("\nVerify labelled values match:")
print(new_series.equals(valid_scores))
```
