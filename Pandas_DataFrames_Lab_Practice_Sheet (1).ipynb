```python
# Question 27
data = {
    'Place': ['Delhi', 'Agra', 'Jaipur', 'Mumbai', 'Pune'],
    'Distance': [50, 100, 200, 300, 400]
}

travel_df = pd.DataFrame(data)

# Add Code column
travel_df['Code'] = np.arange(5)

print("After adding Code:")
print(travel_df)

# Create a partially indexed Series
code_series = pd.Series(
    [101, 104],
    index=[1, 4]
)

# Replace Code column
travel_df['Code'] = code_series

print("After replacing Code:")
print(travel_df)
```

```python
# Question 28
data = {
    'Student': ['Aman', 'Riya', 'Rahul', 'Priya', 'Neha'],
    'Class': [10, 10, 12, 11, 12],
    'Section': ['A', 'B', 'A', 'C', 'B']
}

school_df = pd.DataFrame(data)

new_df = school_df.reindex(
    columns=['Student', 'Class', 'Section', 'House']
)

print("Before assigning House:\n",new_df)

new_df['House'] = 'Blue House'

print("\n After assigning House: \n",new_df)
```

```python
# Question 29
data = {
    'Item': ['Pen', 'Notebook', 'Pencil', 'Eraser', 'Scale'],
    'Quantity': [10, 20, 15, 12, 8],
    'Price': [5, 40, 10, 5, 15]
}

inventory_df = pd.DataFrame(data)

stock_series = pd.Series(
    [100, 200],
    index=[1, 3])

inventory_df['Stock'] = stock_series

print("After adding Stock:\n ",inventory_df)

inventory_df = inventory_df.drop('Stock', axis=1)

print("\n After deleting Stock:\n ",inventory_df)
```
