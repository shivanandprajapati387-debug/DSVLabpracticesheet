```python
# Question 7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)


x = 6
y = 3

swap_message(x, y)

print("Outside function:")
print("x =", x)
print("y =", y)
```
