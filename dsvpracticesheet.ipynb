```python
import pandas as pd
from pandas import Series

# Create Series
sales = Series(
    [450, 750, 300, 900],
    index=["Pen", "Notebook", "Bag", "Laptop"]
)

# Display values
print("Values:")
print(sales.values)

# Display index
print("\nIndex:")
print(sales.index)

# Select one label
print("\nLaptop sales:")
print(sales["Laptop"])

# Filter sales above 500
print("\nSales above 500:")
print(sales[sales > 500])

# Check whether label exists
print("\nIs Bag present?")
print("Bag" in sales.index)

# Convert to dictionary
sales_dict = sales.to_dict()

print("\nDictionary:")
print(sales_dict)
```

```python
data = {
    "A": 15,
    "B": 35,
    "C": 25
}

s = Series(
    data,
    index=["C", "A", "D", "B"]
)

s.name = "Observations"
s.index.name = "Category"
print(s[s > 20])
print(pd.notnull(s))
```
